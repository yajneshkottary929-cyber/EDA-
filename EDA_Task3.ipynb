```python
df["Price_delta"]=df["Close"]-df["Open"]
```

```python
df["Daily_Return"]=(df["Close"]-df["Open"]/df["Open"])*100
```

```python
df["Volume_MA_7"]=df["Volume"].rolling(7).mean()
```

```python
Volume_mean=df["Volume"].mean()
Volume_mean
```

```python
Volume_std=df["Volume"].std()
Volume_std
```

```python
Volume_threshol=Volume_mean+(2*Volume_std)
```

```python
anomalies=df[df["Volume"]>Volume_threshol]
anomalies
```

```python
mean_return=df["Daily_Return"].mean()
variance_return=df["Daily_Return"].var()
standard_return=df["Daily_Return"].std()
```

```python
print("Daily_Return Statistics")
print("mean_Return",mean_return)
print("variance_return",variance_return)
print("standard_return",standard_return)
```
