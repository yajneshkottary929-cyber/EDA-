```python
categarical_columns=["gender","race/ethnicity","parental level of education","lunch","test preparation course","math score","reading score","writing score"]
```

```python
for column in categarical_columns:
    df[column]=df[column].astype(str).str.strip().str.lower()
```

```python
df[column]
```

```python
score_columns=["math score","reading score","writing score"]
```

```python
for columns in score_columns:
    mean=df[columns].mean()
    median=df[columns].median()
    mode=(df[columns].mode()[0])  
```

```python
print("mean_score",mean)
print("median_score",median)
print("mode_score",mode)
```

```python
for columns in score_columns:
    mean=df[columns].mean()
    median=df[columns].median()
    mode=df[columns].mode()
    std=df[columns].std()
    Q1=df[columns].quantile(0.25)
    Q3=df[columns].quantile(0.75)
```

```python
print("mean_score",mean)
print("median_score",median)
print("mode_score",mode)
print("Q1",Q1)
print("Q3",Q3)
```

```python
for columns in ["math score", "reading score", "writing score"]:
    IQR=Q3-Q1
    Lower=Q1-1.5*IQR
    Upper=Q3+1.5*IQR
    Outliers=df[(df[columns]<Lower) | 
               (df[columns]>Upper)]
    
    print("outliers in", columns)
    print(Outliers[columns])
```

```python
performance = []
for Percentage in df["Percentage"]:
    if Percentage >= 90:
        performance.append("Excellent")
    elif Percentage >= 75:
        performance.append("Very good")
    elif Percentage >= 60:
        performance.append("good")
    elif Percentage >= 50:
        performance.append("Average")
    else:
        performance.append("need a improment")

df["performance"] = performance
print(df[["performance","performance"]])
```

```python
df["Total marks"]=df[score_columns].sum(axis=1)
```

```python
df["Percentage"]=(df["Total marks"]/300)*100
```

```python
print(df[score_columns+["Total marks","Percentage"]])
```

```python
df.to_csv("cleaned_student_performance.csv",index=False)
```
