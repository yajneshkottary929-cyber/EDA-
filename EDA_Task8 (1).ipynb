```python
parental_avg=df.groupby("parental level of education")[
    ["math score", "reading score", "writing score"]].mean()

print("\nAverage scores by test preparation:")
print(parental_avg)
```

```python
parental_avg.plot(
    kind="bar",
    figsize=(12,7)
)

plt.title("Average Test Scores by Parental Education Level")
plt.xlabel("Parental Education Level")
plt.ylabel("Average Score")
plt.xticks(rotation=45)
plt.legend(title="Subject")
plt.tight_layout()
plt.show()
```

```python
lunch_avg=df.groupby("lunch")[
    ["math score","reading score","writing score"]].mean()

print("/Average lunch score")
print(lunch_avg)
```

```python
lunch_avg.plot(
    kind="bar",
    figsize=(12,7)
)

plt.title("Average Test Scores by lunch")
plt.xlabel("lunch")
plt.ylabel("Average Score")
plt.xticks(rotation=45)
plt.legend()
plt.tight_layout()
plt.show()
```

```python
sns.boxplot(data=df,x="test preparation course",y="math score")
plt.title("Math Score by test preparation")
plt.xlabel("Test preparation")
plt.ylabel("Math Score")
plt.tight_layout()
plt.show()
```

```python
sns.boxplot(data=df[df["test preparation course"]=="none"],y="math score")
plt.title("Test preparation who did not")
plt.xlabel("Test preparation course")
plt.ylabel("none")
plt.tight_layout()
plt.show()
```

```python
sns.scatterplot(x="math score",y="reading score",data=df)
plt.title("Math Score vs reading score")
plt.xlabel("Math Score")
plt.ylabel("reading Score")
plt.grid(True)
plt.show()
```

```python
sns.scatterplot(x="math score",y="writing score",data=df)
plt.title("math Score vs writing score")
plt.xlabel("math Score")
plt.ylabel("writing Score")
plt.grid(True)
plt.show()
```

```python
sns.scatterplot(x="reading score",y="writing score",data=df)
plt.title("reading Score vs writing score")
plt.xlabel("reading Score")
plt.ylabel("writing Score")
plt.grid(True)
plt.show()
```
