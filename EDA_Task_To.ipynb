```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df=pd.read_csv("profit_sales_analysis_dataset.csv")
```

```python
df
```

```python
sns.histplot(df["Sales"],bins=8)
plt.title("distribution of sales")
plt.show()
```

```python
sns.histplot(df["Profit"],bins=8)
plt.title("distribution of Profit")
plt.show()
```

```python
sns.boxplot (x=df["Sales"])
plt.title("boxplot of Sales")
plt.show()            
```

```python
sns.boxplot (x=df["Profit"])
plt.title("boxplot of Profit")
plt.show()
```

```python
sns.scatterplot (x="Discount_Percent",y="Profit_Margin_Percent",data=df)
plt.title("Discount_Percent vs Profit_Margin_Percent")
plt.show()
```

```python
sns.barplot(x="Discount_Percent",y="Profit_Margin_Percent",data=df)
plt.title("Discount_Percent vs Profit_Margin_Percent")
plt.show()
```

```python
numeric_cols = [
    "Transaction_ID",
    "Unit_Price",
    "Quantity",
    "Discount_Percent",
    "Sales",
    "Profit",
    "Profit_Margin_Percent",
    "Cost",
]

correlation=df[numeric_cols].corr()
```

```python
sns.heatmap(correlation,
            annot=True,      
            cmap="Greens",   
            cbar=False,      
            square=True,    
            linewidth=7)   
plt.show()
```
